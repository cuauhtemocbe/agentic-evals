## Introducción a pytest

`pytest` es un *framework* de pruebas robusto y fácil de usar para Python. Se utiliza para escribir pruebas unitarias, de integración y funcionales. Algunas de sus principales ventajas incluyen:

*   **Sintaxis sencilla**: Las pruebas se escriben como funciones Python estándar, lo que las hace fáciles de leer y mantener.
*   **Detección automática de pruebas**: `pytest` puede descubrir automáticamente las pruebas en tus archivos.
*   **Funcionalidades avanzadas**: Ofrece características como *fixtures* (para configurar y desmontar entornos de prueba), *parametrización* (para ejecutar la misma prueba con diferentes conjuntos de datos) y *plugins*.
*   **Informes detallados**: Proporciona una salida clara y concisa de los resultados de las pruebas.

### Estructura de tests reutilizables con `pytest`

Para estructurar tests reutilizables, `pytest` anima a organizar el código de la aplicación y el código de prueba en directorios separados. Los archivos de prueba deben comenzar con `test_` o terminar con `_test.py`. Las funciones de prueba deben comenzar con `test_`.

Un concepto clave para la reutilización son los `fixtures`. Los *fixtures* son funciones que `pytest` ejecuta antes (y opcionalmente después) de tus tests para configurar un entorno de prueba. Son ideales para inicializar bases de datos, cargar datos, crear clientes de API, etc.

A continuación, crearemos un archivo de Python llamado `my_app.py` que contiene una función simple que queremos probar. Luego, crearemos un archivo de prueba llamado `test_my_app.py` que contendrá las pruebas para nuestra aplicación. **Ten en cuenta que estos archivos se crearán en el sistema de archivos temporal de Colab.**

```python
%%writefile my_app.py

def add(a, b):
    """Suma dos números."""
    return a + b

def multiply(a, b):
    """Multiplica dos números."""
    return a * b
```

```python
%%writefile test_my_app.py
import pytest
from my_app import add, multiply

# Un fixture simple para configurar un valor
@pytest.fixture
def default_numbers():
    return 5, 3

def test_add_function():
    assert add(1, 2) == 3
    assert add(-1, 1) == 0
    assert add(0, 0) == 0

def test_multiply_function():
    assert multiply(2, 3) == 6
    assert multiply(-1, 5) == -5
    assert multiply(0, 10) == 0

def test_add_with_fixture(default_numbers):
    num1, num2 = default_numbers
    assert add(num1, num2) == 8

def test_multiply_with_fixture(default_numbers):
    num1, num2 = default_numbers
    assert multiply(num1, num2) == 15
```

Para ejecutar las pruebas, primero necesitamos instalar `pytest`.

Ahora, podemos ejecutar `pytest` desde la línea de comandos en Colab. `pytest` buscará automáticamente los archivos de prueba en el directorio actual (o subdirectorios) que sigan las convenciones de nombres.

```python
!pytest
```

### Automatización de ejecuciones de evaluación

Para automatizar las ejecuciones de pruebas, generalmente se integra `pytest` en un sistema de Integración Continua (CI) como GitHub Actions, GitLab CI, Jenkins, etc. En Colab, la automatización es más limitada, pero puedes ejecutar los tests mediante comandos de shell.

Si quisieras integrar esto en un *script* de automatización o un flujo de CI, simplemente llamarías al comando `pytest`.

También puedes pasar argumentos a `pytest` para personalizar su comportamiento:

*   `-v`: Modo verboso, muestra más detalles sobre las pruebas.
*   `-s`: Permite la salida de `print` de las pruebas.
*   `-k "nombre_de_la_prueba"`: Ejecuta solo las pruebas que coinciden con el patrón.
*   `--html=report.html`: Genera un informe HTML.
*   `--disable-warnings`: Deshabilita los warnings de `pytest`.

Ejecutemos `pytest` en modo verboso:

```python
!pytest -v
```

Ejecutemos solo las pruebas que contienen la palabra "add":

```python
!pytest -k "add"
```
