### Probando Pandas y Matplotlib

# este es un texto de prueba

```python
import pandas as pd
import matplotlib.pyplot as plt

print('Pandas y Matplotlib importados correctamente.')
```

```python
# Crear un DataFrame simple con pandas
data = {'Nombres': ['Alice', 'Bob', 'Charlie', 'David'],
        'Edades': [24, 27, 22, 32]}
df = pd.DataFrame(data)

print('DataFrame creado:')
display(df)
```

```python
# Crear un gráfico simple con matplotlib
plt.figure(figsize=(6, 4))
plt.bar(df['Nombres'], df['Edades'], color='skyblue')
plt.xlabel('Nombres')
plt.ylabel('Edades')
plt.title('Edades de Personas')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

print('Gráfico generado exitosamente.')
```
