### 1. Validación de Formato (JSON)

La validación de JSON es crucial para asegurar que los datos recibidos o generados se ajusten a un formato estructurado. Un error común es una sintaxis JSON inválida. Utilizaremos el módulo `json` de Python para demostrarlo.

```python
import json

# Ejemplo de JSON inválido
invalid_json_string = "{'nombre': 'Ana', 'edad': 30}" # Usa comillas simples en lugar de dobles

try:
    json.loads(invalid_json_string)
    print("JSON válido (error no detectado)")
except json.JSONDecodeError as e:
    print(f"Error al decodificar JSON (inválido): {e}")

# Ejemplo de JSON válido
valid_json_string = '{"nombre": "Ana", "edad": 30}'

try:
    data = json.loads(valid_json_string)
    print(f"JSON válido: {data}")
except json.JSONDecodeError as e:
    print(f"Error al decodificar JSON (válido): {e}")
```

### 2. Expresiones Regulares (Regex)

Las expresiones regulares son poderosas para buscar patrones de texto. Los errores suelen ocurrir cuando el patrón no coincide con el formato esperado o cuando el patrón es incorrecto. Usaremos el módulo `re`.

```python
import re

# Texto de ejemplo
text = "Mi correo es usuario@ejemplo.com y mi teléfono es 123-456-7890."

# Patrón Regex incorrecto para correos (le falta el punto en .com)
incorrect_email_pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\b"
match = re.search(incorrect_email_pattern, text)
if match:
    print(f"Coincidencia de correo (patrón incorrecto): {match.group(0)}")
else:
    print("No se encontró coincidencia para correo (patrón incorrecto).")

# Patrón Regex correcto para correos
correct_email_pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b"
match = re.search(correct_email_pattern, text)
if match:
    print(f"Coincidencia de correo (patrón correcto): {match.group(0)}")
else:
    print("No se encontró coincidencia para correo (patrón correcto).")

# Patrón Regex para teléfono
phone_pattern = r"\d{3}-\d{3}-\d{4}"
match = re.search(phone_pattern, text)
if match:
    print(f"Coincidencia de teléfono: {match.group(0)}")
else:
    print("No se encontró coincidencia para teléfono.")
```

### 3. Restricciones Físicas (Longitud y Conteo de Palabras)

Las restricciones de longitud o conteo de palabras son comunes en campos de texto, como nombres de usuario, contraseñas o descripciones. Un error ocurre cuando el texto excede o no cumple con los límites definidos.

```python
# Ejemplo de texto que excede la longitud máxima
long_text = "Este es un texto muy largo que excede la longitud máxima permitida para este campo. Debe ser acortado."
max_length = 50

if len(long_text) > max_length:
    print(f"Error: El texto excede la longitud máxima de {max_length} caracteres. Longitud actual: {len(long_text)}")
    # Resolución: Acortar el texto
    truncated_text = long_text[:max_length]
    print(f"Texto acortado: {truncated_text}...")
else:
    print(f"Texto dentro del límite de longitud: {long_text}")

# Ejemplo de texto con conteo de palabras fuera de rango
short_sentence = "Hola mundo."
min_words = 5

words = short_sentence.split()
if len(words) < min_words:
    print(f"Error: El texto tiene menos de {min_words} palabras. Conteo actual: {len(words)}")
    # Resolución: Pedir al usuario que añada más contenido
    print("Por favor, añade más palabras para cumplir con el mínimo requerido.")
else:
    print(f"Texto cumple con el conteo de palabras: {short_sentence}")
```

### 4. Simulación de Llamadas a API (Errores y Corrección)

Las llamadas a APIs son fundamentales para la integración entre sistemas. Es común encontrar errores debido a problemas de red, URLs incorrectas, autenticación fallida o parámetros inválidos. Vamos a simular un escenario de error y luego su corrección.

```python
import requests

# --- Simulación de una API con error ---

# URL de API incorrecta o que no existe
incorrect_api_url = "https://api.ejemplo.com/datos/invalido"

print("--- Intentando llamar a una API con URL incorrecta ---")
try:
    response = requests.get(incorrect_api_url, timeout=5)
    response.raise_for_status() # Lanza una excepción para códigos de estado HTTP de error (4xx o 5xx)
    print(f"Respuesta exitosa (inesperado): {response.json()}")
except requests.exceptions.HTTPError as e:
    print(f"Error HTTP: {e.response.status_code} - {e.response.reason}")
except requests.exceptions.ConnectionError as e:
    print(f"Error de Conexión: No se pudo conectar a la URL '{incorrect_api_url}'. Verifica la URL o tu conexión a internet.")
except requests.exceptions.Timeout as e:
    print(f"Error de Timeout: La solicitud a la URL '{incorrect_api_url}' excedió el tiempo límite.")
except requests.exceptions.RequestException as e:
    print(f"Error de Solicitud General: {e}")
```

### Explicación del Error y Corrección

En el ejemplo anterior, la llamada falló con un `Error de Conexión` (o similar, dependiendo de la disponibilidad de `api.ejemplo.com`). Esto se debe a que la URL proporcionada no es una API real o no es accesible.

Para corregirlo, necesitamos una URL de API válida y posiblemente un manejo adecuado de autenticación o parámetros. Usaremos una API pública de ejemplo para demostrar una llamada exitosa.

```python
# --- Simulación de una API corregida y exitosa ---

# URL de una API pública válida (ejemplo: JSONPlaceholder para posts)
correct_api_url = "https://jsonplaceholder.typicode.com/posts/1"

print("--- Intentando llamar a una API con URL correcta ---")
try:
    response = requests.get(correct_api_url, timeout=5)
    response.raise_for_status() # Lanza una excepción para códigos de estado HTTP de error (4xx o 5xx)
    data = response.json()
    print(f"Llamada a API exitosa. Código de estado: {response.status_code}")
    print("Datos recibidos:")
    print(json.dumps(data, indent=2))
except requests.exceptions.HTTPError as e:
    print(f"Error HTTP: {e.response.status_code} - {e.response.reason}")
except requests.exceptions.ConnectionError as e:
    print(f"Error de Conexión: No se pudo conectar a la URL '{correct_api_url}'.")
except requests.exceptions.Timeout as e:
    print(f"Error de Timeout: La solicitud a la URL '{correct_api_url}' excedió el tiempo límite.")
except requests.exceptions.RequestException as e:
    print(f"Error de Solicitud General: {e}")
```
